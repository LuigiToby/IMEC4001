### d) Modos espaciales

Los modos espaciales obtenidos en el numeral anterior son

$$
(UV)_{mn}(x,y)
=\sin\left(\frac{n\pi x}{L}\right)
\sin\left(\frac{m\pi y}{L}\right),
$$

para

$$
m,n=1,2,3.
$$

Por tanto, se tienen $3\times3=9$ modos espaciales.

### Graficación de los modos

Se grafican los nueve modos espaciales $(UV)_{mn}(x,y)$ para $m,n=1,2,3$ utilizando una superficie tridimensional.

```python
import numpy as np
import matplotlib.pyplot as plt

L = 1

x = np.linspace(0, L, 100)
y = np.linspace(0, L, 100)

X, Y = np.meshgrid(x, y)

for m in range(1, 4):
    for n in range(1, 4):

        UV = np.sin(n*np.pi*X/L) * np.sin(m*np.pi*Y/L)

        fig = plt.figure(figsize=(7, 5))
        ax = fig.add_subplot(111, projection="3d")

        ax.plot_surface(X, Y, UV)

        ax.set_title(fr"$(UV)_{{{m}{n}}}(x,y)$")
        ax.set_xlabel("$x$")
        ax.set_ylabel("$y$")
        ax.set_zlabel("$UV$")

        plt.show()
```

Las gráficas muestran los nueve modos correspondientes a las combinaciones $m,n=1,2,3$. Se observa que $n$ determina el número de oscilaciones en la dirección $x$, mientras que $m$ determina el número de oscilaciones en la dirección $y$.
