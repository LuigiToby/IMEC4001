#### 1. Planteamiento del sistema y valores de $c$

## 1. El problema expuesto en clase

Se pide estudiar

$$
\frac{d}{dt}\begin{pmatrix}x\\y\end{pmatrix}
=\begin{pmatrix}F_1\\F_2\end{pmatrix}
=\begin{pmatrix}x(y-y_0)\\-y\big((x-x_0)+c(y-y_0)\big)\end{pmatrix},
\qquad x_0=y_0=1,
$$

para

$$c\in\left\{-\frac12,\;0,\;\frac12,\;2\right\}.$$

Uso $$\mathbf F=(F_1,F_2)^T$$. Los puntos de equilibrio son los ceros de $$\mathbf F$$. Para clasificarlos evalúo el Jacobiano $$A$$ en cada cero y uso el diagrama de Poincaré en las coordenadas $$\operatorname{tr}A$$ y $$\det A$$.

#### 1(a)–(c). Ceros, clasificación y direcciones principales

### 1(a). ¿Cuáles son los ceros de $$(F_1,F_2)$$?

De $$F_1=x(y-1)=0$$ salen las ramas $$x=0$$ o $$y=1$$.

- Si $$y=1$$, $$F_2=-(x-1)=0$$, luego $$(x,y)=(1,1)$$.
- Si $$x=0$$, $$F_2=-y[-1+c(y-1)]=0$$. Se obtiene $$y=0$$ o, si $$c\ne0$$, $$y=1+1/c$$.

Al sustituir los valores pedidos:

| $$c$$ | Ceros de $$\mathbf F$$ |
|--:|:--|
| $$-1/2$$ | $$(0,0),\;(0,-1),\;(1,1)$$ |
| $$0$$ | $$(0,0),\;(1,1)$$ |
| $$1/2$$ | $$(0,0),\;(0,3),\;(1,1)$$ |
| $$2$$ | $$(0,0),\;(0,3/2),\;(1,1)$$ |

Esta lista responde el inciso (a); en los incisos siguientes los denoto por $$P_0=(0,0)$$, $$P_c=(0,1+1/c)$$ cuando existe, y $$P_*=(1,1)$$.

### Jacobiano que uso en 1(b)–(d)

Derivando las dos componentes una sola vez,

$$
A(x,y)=D\mathbf F(x,y)
=\begin{pmatrix}
y-1 & x\\
-y & -(x-1)-c(2y-1)
\end{pmatrix}.
$$

En un equilibrio calculo

$$\tau=\operatorname{tr}A,\qquad \Delta=\det A,\qquad D=\tau^2-4\Delta.$$

En el diagrama de Poincaré, $$\Delta<0$$ indica silla; para $$\Delta>0$$, $$D<0$$ da espiral (o centro si $$\tau=0$$), y $$D>0$$ da nodo. El signo de $$\tau$$ distingue sumidero de fuente. Si $$D=0$$, hay que revisar los autovectores.

### 1(b). Clasificación de cada cero

En $$P_0$$ los autovalores son $$-1$$ y $$1+c$$, así que el origen es silla en los cuatro valores. En $$P_c$$, la matriz es triangular y sus autovalores son $$1/c$$ y $$-(1+c)$$. En $$P_*$$,

$$A(P_*)=\begin{pmatrix}0&1\\-1&-c\end{pmatrix},\qquad
\lambda^2+c\lambda+1=0.$$

| $$c$$ | Equilibrio | $$\operatorname{tr}A$$ | $$\det A$$ | Autovalores | Clasificación |
|--:|:--|--:|--:|:--|:--|
| $$-1/2$$ | $$(0,0)$$ | $$-1/2$$ | $$-1/2$$ | $$-1,\;1/2$$ | silla |
| $$-1/2$$ | $$(0,-1)$$ | $$-5/2$$ | $$1$$ | $$-2,\;-1/2$$ | atractor, nodo sumidero |
| $$-1/2$$ | $$(1,1)$$ | $$1/2$$ | $$1$$ | $$1/4\pm i\sqrt{15}/4$$ | repulsor, espiral fuente |
| $$0$$ | $$(0,0)$$ | $$0$$ | $$-1$$ | $$-1,\;1$$ | silla |
| $$0$$ | $$(1,1)$$ | $$0$$ | $$1$$ | $$\pm i$$ | elipse (centro) |
| $$1/2$$ | $$(0,0)$$ | $$1/2$$ | $$-3/2$$ | $$-1,\;3/2$$ | silla |
| $$1/2$$ | $$(0,3)$$ | $$-1$$ | $$-3$$ | $$2,\;-3/2$$ | silla |
| $$1/2$$ | $$(1,1)$$ | $$-1/2$$ | $$1$$ | $$-1/4\pm i\sqrt{15}/4$$ | atractor, espiral sumidero |
| $$2$$ | $$(0,0)$$ | $$2$$ | $$-3$$ | $$-1,\;3$$ | silla |
| $$2$$ | $$(0,3/2)$$ | $$-5/2$$ | $$-3/2$$ | $$1/2,\;-3$$ | silla |
| $$2$$ | $$(1,1)$$ | $$-2$$ | $$1$$ | $$-1\;(\text{doble})$$ | atractor, nodo degenerado |

Para $$c=0$$, el Jacobiano por sí solo no distingue un centro de una espiral con parte real cero. Aquí sí es un centro no lineal: en $$x,y>0$$,

$$H(x,y)=x-\ln x+y-\ln y$$

se conserva a lo largo de las trayectorias, pues $$dH/dt=0$$. Sus curvas de nivel cerradas alrededor de $$(1,1)$$ confirman la clasificación como elipse (centro), estable pero no atractor.

Para $$c=2$$, el autovalor doble $$-1$$ tiene un solo autovector independiente; por eso el punto es nodo degenerado y atractor.

### Diagrama de Poincaré

Cada marcador corresponde a un equilibrio de la tabla. La parábola $$\Delta=\tfrac14(\operatorname{tr}A)^2$$ separa nodos y espirales; las sillas quedan bajo el eje $$\Delta=0$$.

```python
import numpy as np
import matplotlib.pyplot as plt

casos_c = [-0.5, 0.0, 0.5, 2.0]
colores_c = {-0.5: "#c43c39", 0.0: "#26828e", 0.5: "#e69f00", 2.0: "#5555a5"}

def F(x, y, c):
    return x*(y-1), -y*((x-1) + c*(y-1))

def A_en(x, y, c):
    return np.array([[y-1, x], [-y, -(x-1)-c*(2*y-1)]], dtype=float)

def puntos(c):
    ps = [(0., 0., r"$P_0=(0,0)$"), (1., 1., r"$P_*=(1,1)$")]
    if c != 0:
        ps.insert(1, (0., 1+1/c, rf"$P_c=(0,{1+1/c:g})$"))
    return ps

fig, ax = plt.subplots(figsize=(8.4, 5.8), constrained_layout=True)
t = np.linspace(-3.1, 3.1, 500)
ax.plot(t, t*t/4, color="black", lw=1.5, label=r"$D=0:\ \Delta=\tau^2/4$")
ax.axhline(0, color="0.3", lw=.8); ax.axvline(0, color="0.3", lw=.8)
for c in casos_c:
    for xe, ye, name in puntos(c):
        J = A_en(xe, ye, c)
        tau, delta = np.trace(J), np.linalg.det(J)
        ax.scatter(tau, delta, color=colores_c[c], s=65, edgecolor="black", zorder=3)
        equilibrium = r"P_0" if np.allclose([xe,ye],[0,0]) else (r"P_*" if np.allclose([xe,ye],[1,1]) else r"P_c")
        offsets = {
            (-.5, r"P_c"): (-34,-20), (2., r"P_*"): (2,13),
            (.5, r"P_*"): (-32,13), (0., r"P_*"): (5,-21), (-.5, r"P_*"): (6,13),
            (-.5, r"P_0"): (-25,-19), (0., r"P_0"): (5,-19),
            (.5, r"P_0"): (6,-19), (2., r"P_0"): (6,-19),
            (.5, r"P_c"): (6,5), (2., r"P_c"): (-30,-18),
        }
        ax.annotate(rf"${equilibrium}$", (tau, delta), xytext=offsets[(c,equilibrium)],
                    textcoords="offset points", fontsize=8)
    ax.scatter([], [], color=colores_c[c], edgecolor="black", label=rf"$c={c:g}$")
ax.set(xlim=(-3.2,3.2), ylim=(-3.5,2.7), xlabel=r"$\operatorname{tr}A$", ylabel=r"$\det A$",
       title="Diagrama de Poincaré de los equilibrios")
ax.grid(alpha=.23); ax.legend()
plt.show()
```

### 1(c). Direcciones principales cerca de cada cero

Para una silla, la dirección propia asociada al autovalor negativo es estable y la asociada al positivo es inestable. En un nodo sumidero las dos direcciones son estables; en las espirales las trayectorias entran o salen según el signo de la parte real. En el centro, las trayectorias cerradas no se acercan ni se alejan del equilibrio.

En cada panel trazo el campo lineal $$\dot{\boldsymbol\xi}=A(P)\boldsymbol\xi$$ alrededor del equilibrio, donde $$\boldsymbol\xi=(x-x_P,y-y_P)^T$$. Las líneas verde y roja son las direcciones propias estables e inestables. Para $$c=0$$ en $$(1,1)$$ uso el campo no lineal y se ven las órbitas cerradas.

```python
def dibuja_locales(c):
    ps = puntos(c)
    fig, axes = plt.subplots(1, len(ps), figsize=(4.1*len(ps), 4.0), constrained_layout=True)
    if len(ps) == 1: axes = [axes]
    q = np.linspace(-.65, .65, 33)
    QX, QY = np.meshgrid(q, q)
    for ax, (xe, ye, name) in zip(axes, ps):
        J = A_en(xe, ye, c)
        if c == 0 and np.allclose([xe,ye], [1,1]):
            U, V = F(xe+QX, ye+QY, c)
        else:
            U, V = J[0,0]*QX + J[0,1]*QY, J[1,0]*QX + J[1,1]*QY
        ax.streamplot(q, q, U, V, density=1.0, color="#457b9d", linewidth=.9, arrowsize=1.0)
        vals, vecs = np.linalg.eig(J)
        for lam, vec in zip(vals, vecs.T):
            if abs(lam.imag) < 1e-8:
                vec = np.real(vec); vec /= np.linalg.norm(vec)
                color = "#2a9d8f" if lam.real < 0 else "#e63946"
                for s in (-1,1):
                    start = s*.52*vec
                    end = s*.30*vec if lam.real < 0 else s*.68*vec
                    ax.annotate("", xy=end, xytext=start,
                                arrowprops={"arrowstyle":"-|>","color":color,"lw":1.5})
        ax.scatter(0,0,s=36,color="black",zorder=4)
        ax.set(xlim=(-.7,.7),ylim=(-.7,.7),aspect="equal",xlabel=r"$x-x_P$",ylabel=r"$y-y_P$",
               title=f"{name}\n{np.linalg.eigvals(J)[0]:.2g}, {np.linalg.eigvals(J)[1]:.2g}")
        ax.axhline(0,color="0.7",lw=.5); ax.axvline(0,color="0.7",lw=.5); ax.grid(alpha=.15)
    fig.suptitle(rf"Direcciones locales, $c={c:g}$  (verde: estable; rojo: inestable)")
    plt.show()

for c in casos_c:
    dibuja_locales(c)
```

#### 1(d)–(f). Direcciones $U_R,U_I$, retratos y caso $c=1$

### 1(d). Direcciones $$U_R,U_I$$ para elipse y espirales

En el equilibrio $$(1,1)$$, para un par propio $$\lambda=\alpha+i\beta$$ tomo el autovector $$\mathbf q=(1,\lambda)^T=U_R+iU_I$$, siguiendo la notación de la Tarea 7. Así,

$$U_R=\begin{pmatrix}1\\\alpha\end{pmatrix},\qquad
U_I=\begin{pmatrix}0\\\beta\end{pmatrix},$$

y las soluciones reales asociadas son proporcionales a

$$e^{\alpha t}(U_R\cos\beta t-U_I\sin\beta t),\qquad
e^{\alpha t}(U_R\sin\beta t+U_I\cos\beta t).$$

Para $$c=\pm1/2$$, $$\alpha=-c/2$$ y $$\beta=\sqrt{4-c^2}/2$$. Para $$c=0$$, $$\alpha=0,\beta=1$$; se obtiene la elipse (centro). Los vectores se dibujan desde $$(1,1)$$ en el siguiente gráfico.

```python
fig, axes = plt.subplots(1, 3, figsize=(11,3.8), constrained_layout=True)
for ax, c in zip(axes, [-.5, 0., .5]):
    alpha, beta = -c/2, np.sqrt(4-c*c)/2
    UR = np.array([1., alpha]); UI = np.array([0., beta]); base=np.array([1.,1.])
    ax.quiver(*base,*UR,angles="xy",scale_units="xy",scale=1,color="#2878b5",label=r"$U_R$")
    ax.quiver(*base,*UI,angles="xy",scale_units="xy",scale=1,color="#e07a32",label=r"$U_I$")
    ax.scatter(*base,color="black",s=25,zorder=3)
    ax.set(xlim=(-.2,2.3),ylim=(-.8,2.2),aspect="equal",xlabel="$x$",ylabel="$y$",title=rf"$c={c:g}$")
    ax.grid(alpha=.2); ax.legend()
fig.suptitle(r"Partes real e imaginaria del autovector en $(1,1)$")
plt.show()
```

### 1(e). Retrato global para cada caso

Con los retratos locales anteriores, el siguiente `streamplot` reúne el flujo de cada sistema y marca todos sus ceros. Reemplazo el dibujo a mano alzada por el mismo esquema generado con el campo de velocidad.

```python
def retrato_global(c):
    x=np.linspace(-2.0,2.8,65); y=np.linspace(-2.3,4.0,75)
    X,Y=np.meshgrid(x,y); U,V=F(X,Y,c)
    fig,ax=plt.subplots(figsize=(6.8,5.3),constrained_layout=True)
    ax.streamplot(x,y,U,V,density=1.15,color="#397a9e",linewidth=.85,arrowsize=1.0)
    for xe,ye,name in puntos(c):
        ax.scatter(xe,ye,s=55,color="#d62828",edgecolor="white",zorder=4)
        ax.annotate(name,(xe,ye),xytext=(6,6),textcoords="offset points",fontsize=8)
    ax.set(xlim=(x.min(),x.max()),ylim=(y.min(),y.max()),xlabel="$x$",ylabel="$y$",
           title=rf"Retrato global, $c={c:g}$")
    ax.axhline(0,color="0.5",lw=.6); ax.axvline(0,color="0.5",lw=.6); ax.grid(alpha=.15)
    plt.show()

for c in casos_c:
    retrato_global(c)
```

#### 2. Vorticidad y velocidad tangencial

## 2. Aparición de vórtices en un flujo

La figura da la rapidez tangencial del vórtice de núcleo rotacional:

$$
V_\theta(r)=\begin{cases}
\omega r,&r<r_0,\\[1mm]
\dfrac{\omega r_0^2}{r},&r>r_0.
\end{cases}
$$

La sumo al flujo uniforme horizontal $$\mathbf U=(U,0)$$. Tomo $$\omega>0$$ y $$U>0$$, como en el esquema; los ceros están sobre el semieje positivo $$y$$, donde la velocidad tangencial apunta contra el flujo uniforme.

#### 2. Flujo uniforme y localización de los ceros

#### 2(a). Valores de $U$ y radios de los ceros

### 2(a). ¿Para qué valores de $$U$$ y en qué radios hay ceros?

Sobre el eje $$y$$, $$x=0$$ y $$r=y$$. La velocidad horizontal total se anula cuando $$U=V_\theta(r)$$. El máximo de $$V_\theta$$ es $$\omega r_0$$ en $$r=r_0$$. Por tanto,

$$
\begin{array}{c|c|c}
\text{condición}&\text{radios de los ceros}&\text{número}\\\hline
U>\omega r_0&\text{ninguno}&0\\
U=\omega r_0&r=r_0&1\\
0<U<\omega r_0&r_1=U/\omega,\quad r_2=\omega r_0^2/U&2
\end{array}
$$

En el último caso, $$r_1<r_0<r_2$$. En el umbral, las dos ramas se encuentran en el borde del núcleo. Si $$U=0$$, existe además el cero en el centro $$r=0$$.

La gráfica siguiente muestra los cruces de la curva $$V_\theta(r)$$ con tres valores de $$U$$. La línea horizontal tangente al máximo corresponde al único cero.

```python
omega, r0 = 1.0, 1.0
r = np.linspace(.001, 3.2, 600)
vtheta = np.where(r<=r0, omega*r, omega*r0**2/r)
fig,ax=plt.subplots(figsize=(7.2,4.4),constrained_layout=True)
ax.plot(r,vtheta,color="#2878b5",lw=2,label=r"$V_\theta(r)$")
for U,color,label in [(1.35,"#d1495b",r"$U>\omega r_0$"),(1.0,"#6a4c93",r"$U=\omega r_0$"),(.55,"#2a9d8f",r"$U<\omega r_0$")]:
    ax.axhline(U,color=color,ls="--",label=label)
    if U<omega*r0:
        ax.scatter([U/omega,omega*r0**2/U],[U,U],color=color,zorder=3)
    elif np.isclose(U,omega*r0): ax.scatter([r0],[U],color=color,zorder=3)
ax.axvline(r0,color="0.3",ls=":",label=r"$r_0$")
ax.set(xlim=(0,3.2),ylim=(0,1.55),xlabel="$r$",ylabel=r"$V_\theta(r)$",title="Intersecciones que determinan los ceros")
ax.grid(alpha=.2);ax.legend()
plt.show()
```

#### 2(b). Velocidad en coordenadas cartesianas

#### 2(b). Gradiente de la velocidad fuera del núcleo

### 2(b). Clasificación de los dos ceros

Para $$x=0$$ y $$y>0$$, las velocidades del enunciado más el flujo uniforme son

$$
\mathbf V(x,y)=\begin{cases}
\begin{pmatrix}U-\omega y\\\omega x\end{pmatrix},&r\le r_0,\\[3mm]
\begin{pmatrix}U-\dfrac{\omega r_0^2y}{r^2}\\[1mm]\dfrac{\omega r_0^2x}{r^2}\end{pmatrix},&r\ge r_0.
\end{cases}
$$

En el caso de dos ceros, $$0<U<\omega r_0$$:

- **Cero interior** $$P_1=(0,r_1)$$, con $$r_1=U/\omega$$. La matriz de velocidad es

$$A_1=\begin{pmatrix}0&-\omega\\\omega&0\end{pmatrix},\qquad
\operatorname{tr}A_1=0,\quad\det A_1=\omega^2.$$

  Sus autovalores son $$\lambda=\pm i\omega$$; es una elipse (centro).

- **Cero exterior** $$P_2=(0,r_2)$$, con $$r_2=\omega r_0^2/U$$. Al evaluar el gradiente del enunciado en $$x=0,y=r_2$$:

$$A_2=\begin{pmatrix}0&q\\q&0\end{pmatrix},\qquad
q=\frac{\omega r_0^2}{r_2^2}>0.$$

  $$\operatorname{tr}A_2=0$$ y $$\det A_2=-q^2<0$$, así que es silla. Los autovalores y autovectores son

$$\lambda_s=-q,\quad \mathbf v_s=\begin{pmatrix}1\\-1\end{pmatrix};\qquad
\lambda_u=q,\quad \mathbf v_u=\begin{pmatrix}1\\1\end{pmatrix}.$$

La dirección $$\mathbf v_s$$ es estable y $$\mathbf v_u$$ inestable. No vuelvo a derivar el gradiente: uso la expresión que aparece en la hoja y la evalúo en el cero exterior.

#### 2(b)–(c). Clasificación y diagramas de flujo

### 2(c). Diagramas aproximados del flujo

En cada panel dibujo las líneas de corriente del campo total y marco los ceros. Se usan $$\omega=r_0=1$$: $$U=1.35$$ (sin ceros), $$U=1$$ (un cero en el borde) y $$U=0.55$$ (dos ceros: centro dentro del núcleo y silla fuera).

La función de corriente que uso satisface $$u=\partial_y\psi$$ y $$v=-\partial_x\psi$$:

$$
\psi(x,y)=Uy-\begin{cases}
\dfrac{\omega r^2}{2},&r\le r_0,\\[1mm]
\omega r_0^2\left[\dfrac12+\ln(r/r_0)\right],&r\ge r_0.
\end{cases}
$$

```python
def velocidad_total(X,Y,U,omega=1.,r0=1.):
    r2=X*X+Y*Y; R=np.sqrt(r2)
    k=np.where(R<=r0,omega,omega*r0**2/np.maximum(r2,1e-14))
    return U-k*Y,k*X

def psi(X,Y,U,omega=1.,r0=1.):
    R=np.hypot(X,Y)
    f=np.where(R<=r0, .5*omega*R**2,
                omega*r0**2*(.5+np.log(np.maximum(R/r0,1e-12))))
    return U*Y-f

def dibuja_vortice(U,titulo):
    x=np.linspace(-3.1,3.1,85);y=np.linspace(-3.1,3.1,85)
    X,Y=np.meshgrid(x,y); VX,VY=velocidad_total(X,Y,U,omega,r0)
    PS=psi(X,Y,U,omega,r0)
    fig,ax=plt.subplots(figsize=(5.6,5.2),constrained_layout=True)
    lev=np.linspace(np.percentile(PS,4),np.percentile(PS,96),28)
    ax.contour(X,Y,PS,levels=lev,colors="#32728f",linewidths=.85)
    ax.streamplot(x,y,VX,VY,color="0.25",density=.85,linewidth=.55,arrowsize=.75)
    ax.add_patch(plt.Circle((0,0),r0,fill=False,ls="--",lw=1.2,color="#e76f51"))
    for rr,region in ([(U/omega,"interior"),(omega*r0**2/U,"exterior")] if U<omega*r0 else ([(r0,"borde")] if np.isclose(U,omega*r0) else [])):
        ax.scatter(0,rr,s=55,color="#d62828",edgecolor="white",zorder=4)
        ax.annotate(rf"$r={rr:.2f}$",(0,rr),xytext=(6,5),textcoords="offset points",fontsize=8)
    ax.set(xlim=(-3.1,3.1),ylim=(-3.1,3.1),aspect="equal",xlabel="$x$",ylabel="$y$",title=titulo)
    ax.axhline(0,color="0.6",lw=.5);ax.axvline(0,color="0.6",lw=.5);ax.grid(alpha=.12)
    plt.show()

for U,titulo in [(1.35,r"No hay ceros: $U$ alto"),(1.0,r"Un cero: $U=\omega r_0$"),(.55,r"Dos ceros: $U$ bajo")]:
    dibuja_vortice(U,titulo)
```

Para $$U=\omega r_0$$ el cero está justo en $$r=r_0$$, donde cambia la fórmula del campo. La velocidad es continua, pero su gradiente no; por eso no asigno una única clasificación lineal a ese caso límite. Para $$U<\omega r_0$$, los dos ceros se separan: el interior es centro y el exterior es silla, como se ve en el tercer diagrama.
